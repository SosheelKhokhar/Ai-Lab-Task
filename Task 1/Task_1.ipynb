```python
def check_scholarship(score, documents_complete, attendance, fee_clearance):
    if score >= 70 and documents_complete and attendance >= 75 and fee_clearance:
        return "Eligible"
    return "Not Eligible"

print("Case 1:", check_scholarship(80, True, 80, True))
print("Case 2:", check_scholarship(85, True, 70, True))
print("Case 3:", check_scholarship(90, True, 90, False))
print("Case 4:", check_scholarship(70, True, 75, True))
```
